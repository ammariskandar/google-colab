```python
# This file is the skeleton base
Y = "BMI Calculator ver 1"
print (Y)
db = []
immutable_info = ("School: IUJ", "Major: DXP")
loop = True
name = input("Please type your name.")
db_name = "Name: " + name.upper()
db.append(db_name)
BMIdata =[]
while loop == True:
  heightstr = input("Please enter your height, in cm.")
  if (float(heightstr)<100):
      print("Input height in cm, not meters or foot")
      continue
  weightstr = input("Please enter your weight, in kg.")
  height = float(heightstr)
  weight = float(weightstr)
  BMI = (weight/(height/100)**2)
  if BMI < 18.5:
    BMIclass = "Underweight"
  elif BMI >= 18.5 and BMI <= 24.9:
    BMIclass = "Normal"
  elif BMI >= 25.0 and 29.9:
    BMIclass = "Overweight"
  else:
    BMIclass = "Obese"

  BMItext = "BMI: " + str(round(BMI,1)) + ". BMI Classification: " + BMIclass
  print (BMItext)
  BMIdata.append (BMItext)
  db = db + list(immutable_info) + BMIdata
  print()
  print()
  print (*db)
  loop = False
```
